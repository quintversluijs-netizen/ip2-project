```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# ==================================================
# 1. MEETGEGEVENS
# ==================================================

Ubron = np.array([
    0.10, 0.12, 0.14, 0.16,
    0.18, 0.20, 0.22, 0.24,
    0.26, 0.28, 0.30, 0.32])

UR = np.array([
    4e-6, 6e-6, 9e-6, 13e-6,
    19e-6, 27e-6, 38e-6, 53e-6,
    73e-6, 100e-6, 136e-6, 183e-6])


# Weerstanden in Ohm
R1 = 10000
R2 = 330
R3 = 10000


# ==================================================
# 2. BEREKENEN VAN DE DIODESPANNING EN STROOM
# ==================================================

# Diodestroom met de wet van Ohm
ID = UR / R1

# Diodespanning, inclusief spanningsval over R2
UD = (R3 / (R2 + R3)) * Ubron - UR


# ==================================================
# 3. CONSTANTEN
# ==================================================

q = 1.602176634e-19
T = 293.15             # Temperatuur in Kelvin

n = 2.0                # Ideality factor
u_n = 0.1              # Onzekerheid idealiteitsfactor


# ==================================================
# 4. SHOCKLEYVERGELIJKING
# ==================================================

def diode(VD, I0, kB):
    return I0 * (np.exp(q * VD / (n * kB * T)) - 1)


# ==================================================
# 5. NIET-LINEAIRE FIT
# ==================================================

popt, pcov = curve_fit(
    diode,
    UD,
    ID,
    p0=[1e-11, 1.3e-23],
    maxfev=100000
)

I0_fit, kB_fit = popt

# Onzekerheden uit de covariantiematrix
u_I0 = np.sqrt(pcov[0, 0])
u_kB_fit = np.sqrt(pcov[1, 1])


# ==================================================
# 6. ONZEKERHEID DOOR DE IDEALITEITSFACTOR
# ==================================================

# Functie om de fit opnieuw uit te voeren voor een andere n
def fit_for_n(n_value):

    def diode_n(VD, I0, kB):
        return I0 * (
            np.exp(q * VD / (n_value * kB * T)) - 1
        )

    parameters, covariance = curve_fit(
        diode_n,
        UD,
        ID,
        p0=[I0_fit, kB_fit],
        maxfev=100000
    )

    return parameters[1]


# Fits voor n - u_n en n + u_n
kB_minus = fit_for_n(n - u_n)
kB_plus = fit_for_n(n + u_n)

# Numerieke afgeleide naar n
dkB_dn = (kB_plus - kB_minus) / (2 * u_n)

# Onzekerheid door idealiteitsfactor
u_kB_n = abs(dkB_dn) * u_n


# ==================================================
# 7. TOTALE ONZEKERHEID
# ==================================================

u_kB_total = np.sqrt(u_kB_fit**2 + u_kB_n**2)


# ==================================================
# 8. RESULTATEN
# ==================================================

print("----- FITRESULTATEN -----")

print(f"I0 = {I0_fit:.3e} ± {u_I0:.3e} A")

print(f"kB uit fit = {kB_fit:.3e} J/K")
print(f"Fitonzekerheid = {u_kB_fit:.3e} J/K")

print(f"Onzekerheid door n = {u_kB_n:.3e} J/K")

print(f"Totale onzekerheid = {u_kB_total:.3e} J/K")

print("\n----- EINDRESULTAAT -----")

print(
    f"kB = ({kB_fit:.2e} ± {u_kB_total:.2e}) J/K"
)


# ==================================================
# 9. STRIJDIGHEIDSANALYSE
# ==================================================

kB_literatuur = 1.380649e-23

S = abs(kB_fit - kB_literatuur) / u_kB_total

print("\n----- STRIJDIGHEIDSANALYSE -----")
print(f"Strijdigheid = {S:.2f} sigma")


# ==================================================
# 10. GRAFIEK
# ==================================================

V_plot = np.linspace(min(UD), max(UD), 500)

plt.figure(figsize=(7, 5))

plt.semilogy(UD, ID, 'o', label='Metingen')
plt.semilogy(V_plot, diode(V_plot, *popt), '-', label='Fit')

plt.xlabel("Diodespanning $U_D$ (V)")
plt.ylabel("Diodestroom $I_D$ (A)")

plt.grid(True, which='both')
plt.legend()
plt.tight_layout()

plt.savefig("Boltzmann grafiek.pdf")
plt.show()

```
